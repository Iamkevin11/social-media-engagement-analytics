# Data Cleaning

```python
# Check missing values using isnull()
print("Missing values in each column:")
print(df.isnull().sum())

# Check missing values using isna()
print("\nTotal missing values:", df.isna().sum().sum())

# Display only columns containing missing values
missing_values = df.isnull().sum()
print("\nColumns with missing values:")
print(missing_values[missing_values > 0])
```

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill missing age values with the median
df["age"] = df["age"].fillna(df["age"].median())

# Fill missing numerical engagement metrics with their medians
numeric_columns = ["likes", "comments", "shares"]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill missing categorical values with the mode
categorical_columns = ["gender", "sentiment"]

for column in categorical_columns:
    mode_value = df[column].mode()[0]
    df[column] = df[column].fillna(mode_value)

# Verify the result
print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nTotal missing values remaining:", df.isnull().sum().sum())
```

```python
# Count duplicate rows
print("Duplicate rows before cleaning:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Verify
print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Dataset shape:", df.shape)
```

```python
# Convert posted_at to datetime
df["posted_at"] = pd.to_datetime(
    df["posted_at"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Check the data type
print(df["posted_at"].dtype)
```

```python
# Clean and standardize categorical columns
text_columns = [
    "gender",
    "country",
    "post_type",
    "post_category",
    "device_type",
    "sentiment"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

# Standardize labels to lowercase
for column in ["country", "post_type", "post_category",
               "device_type", "sentiment"]:
    df[column] = df[column].str.lower()

# Standardize gender labels
df["gender"] = df["gender"].str.title()

# Display unique categories
for column in text_columns:
    print(f"\n{column}:")
    print(sorted(df[column].dropna().unique().tolist()))
```

```python
# Check for negative engagement values
metrics = ["likes", "comments", "shares"]

for column in metrics:
    negative_count = (df[column] < 0).sum()
    print(f"Negative values in {column}: {negative_count}")
```

```python
for column in metrics:
    df.loc[df[column] < 0, column] = np.nan
    df[column] = df[column].fillna(df[column].median())

print("Minimum likes:", df["likes"].min())
print("Minimum comments:", df["comments"].min())
print("Minimum shares:", df["shares"].min())
```

```python
# Count hashtags in each row
df["hashtag_count"] = (
    df["hashtags"]
    .fillna("")
    .str.findall(r"#\w+")
    .str.len()
)

# Display hashtags and their counts
print(df[["hashtags", "hashtag_count"]].head(10))
```

```python
print("Sentiment distribution:")
print(df["sentiment"].value_counts())

print("\nUnique sentiment labels:")
print(df["sentiment"].unique())
```

```python
# Final cleaning report
print("FINAL DATA CLEANING REPORT")
print("-" * 35)

print("Dataset shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
print("Total missing values:", df.isnull().sum().sum())

print("\nData types:")
print(df.dtypes)

print("\nFirst five cleaned rows:")
display(df.head())

print("\nCleaning completed successfully!")
```

```python
# Engagement by age group
age_bins = [0, 18, 25, 35, 45, 60, 100]
age_labels = [
    "Under 18", "18–25", "26–35",
    "36–45", "46–60", "Above 60"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

age_insights = df.groupby("age_group", observed=False).agg(
    total_posts=("age", "count"),
    avg_likes=("likes", "mean"),
    avg_engagement_rate=("engagement_rate", "mean")
).round(2)

print("7. ENGAGEMENT BY AGE GROUP")
display(age_insights)

# Engagement by hour of the day
df["posting_hour"] = df["posted_at"].dt.hour

time_insights = df.groupby("posting_hour").agg(
    total_posts=("posting_hour", "size"),
    avg_impressions=("impression_count", "mean"),
    avg_engagement_rate=("engagement_rate", "mean")
).round(2)

print("8. POSTING HOURS WITH HIGHEST AVERAGE IMPRESSIONS")
display(time_insights.sort_values(
    "avg_impressions", ascending=False
).head(10))
```
